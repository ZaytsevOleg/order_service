```python
def transform_data(data):
    """
    Выполняет трансформацию данных перед загрузкой в MongoDB.
    
    :param data: Исходный список данных (JSON)
    :return: Трансформированный список данных
    """
    # Пример трансформации: переименование ключей и добавление новых полей
    transformed_data = []
    for record in data:
        transformed_record = {
            "order_id": record.get("Ref_Key"),
            "customer": "-",
            "holding": "-",
            "date": record.get("Date"),
            "amount": record.get("СуммаДокумента"),
            "comment": record.get("Комментарий", ""), 
            "brand": "-",           
            "status": 0, 
            "shipping_date":"-",
            "shipping_type":"-",
            "product_list": record.get("Товары")
        }
        transformed_data.append(transformed_record)
    
    print(f"Трансформация завершена. Обработано {len(transformed_data)} записей.")
    return transformed_data



transformed_data = transform_data(raw_data)  # Трансформация данных
```

```python
from pymongo import MongoClient, UpdateOne

def update_documents_in_mongodb(sql_data, unique_field="Ref_key_order"):
    """
    Обновляет документы в MongoDB на основе данных из SQL-запроса.
    
    :param sql_data: DataFrame с данными из SQL-запроса
    :param mongo_uri: URI для подключения к MongoDB
    :param mongo_db: Имя базы данных MongoDB
    :param mongo_collection: Имя коллекции MongoDB
    :param unique_field: Поле для поиска документов в MongoDB (по умолчанию "Ref_key_order")
    """
    if sql_data.empty:
        print("Данные из SQL-запроса пусты, обновление MongoDB не выполнено.")
        return

    try:
        # Подключение к MongoDB
        collection = db.customer_order

        # Преобразование всех дат в datetime.datetime
        if "Date_of_shipment" in sql_data.columns:
            sql_data["Date_of_shipment"] = pd.to_datetime(sql_data["Date_of_shipment"])

        # Подготовка операций обновления
        update_operations = []
        for _, row in sql_data.iterrows():
            # Создаём операцию обновления для каждого документа
            update_operations.append(
                UpdateOne(
                    {"order_id": row[unique_field]},  # Условие поиска
                    {
                        "$set": {
                            "Number": row["Number_Only"],
                            "customer": row["Names"],
                            "holding": row["holding"],
                            "brand": row["name_brand"],
                            "status": row["Status"],
                            "shipping_date": row["Date_of_shipment"],
                            "shipping_type": row["Операция"],
                        }
                    },
                    upsert=False  # Не создавать документ, если он не найден
                )
            )

        # Выполнение операций обновления в одной транзакции (bulk_write)
        if update_operations:
            result = collection.bulk_write(update_operations)
            print(f"Обновлено {result.modified_count} документов.")
        else:
            print("Нет обновлений для выполнения.")
    except Exception as e:
        print(f"Ошибка при обновлении документов в MongoDB: {e}")
    finally:
        client.close()
```
